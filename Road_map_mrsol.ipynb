##### Score mit gezogenen Samples

```python
#Score von schätzung in der logit-transformierten Bereich

predictor = Play.distribution_building.distribution_predictor(Transform=True)
data = Play.shape_data.create_residuals(predictor[0],lat_idx= 16,lon_idx = 32,min_run_idx=5, max_run_idx=11,Transform=True)
res = data.mrsol_res.values.flatten()

tas = data.tas.values.flatten()
tas = tas[np.isfinite(res)]
pr = data.pr.values.flatten()
pr = pr[np.isfinite(res)]
mask = np.isfinite(data.mrsol_res.values)
data["sigma"] = xr.full_like(data["mrsol_res"], np.nan)
data["sigma"].values[mask] = np.sqrt(np.maximum(predictor[1].predict(np.array([tas,pr]).T),1e-6))

z_samples = np.random.normal(
    loc=data.mrsol_est.values[..., None],
    scale=data.sigma.values[..., None],
    size=data.mrsol_est.shape + (1000,)
)


samples_da = xr.DataArray(
    z_samples,
    dims=data.mrsol.dims + ("member",),
    coords={
        **data.mrsol.coords,
        "member": np.arange(z_samples.shape[-1])
    }
)


data_bt = xr.full_like(data, np.nan)
samples_da_bt = xr.full_like(samples_da, np.nan)
data_bt["mrsol"] = 100 / (1 + np.exp(data.mrsol))
samples_da_bt = 100 / (1 + np.exp(samples_da))

score = xs.crps_ensemble(data.mrsol,samples_da)
print(score.values)
score = xs.crps_ensemble(data_bt.mrsol,samples_da_bt)
print(score.values)
```

```python
#Schätzung in Orginal Scala

predictor = Play.distribution_building.distribution_predictor(Transform=False)
data = Play.shape_data.create_residuals(predictor[0],lat_idx= 16,lon_idx = 32,min_run_idx=5, max_run_idx=11,Transform=False)
res = data.mrsol_res.values.flatten()

tas = data.tas.values.flatten()
tas = tas[np.isfinite(res)]
pr = data.pr.values.flatten()
pr = pr[np.isfinite(res)]
mask = np.isfinite(data.mrsol_res.values)
data["sigma"] = xr.full_like(data["mrsol_res"], np.nan)
data["sigma"].values[mask] = np.sqrt(np.maximum(predictor[1].predict(np.array([tas,pr]).T),1e-6))

z_samples = np.random.normal(
    loc=data.mrsol_est.values[..., None],
    scale=data.sigma.values[..., None],
    size=data.mrsol_est.shape + (1000,)
)


samples_da = xr.DataArray(
    z_samples,
    dims=data.mrsol.dims + ("member",),
    coords={
        **data.mrsol.coords,
        "member": np.arange(z_samples.shape[-1])
    }
)

score = xs.crps_ensemble(data.mrsol,samples_da)
score.values
```

```python
#score von starrem mean
mean_da = xr.full_like(samples_da, data.mrsol.mean())
score = xs.crps_ensemble(data.mrsol,mean_da)
score.values
```

##### Score with distribution parameters

```python
#ursprüngliche Scala
predictor = Play.distribution_building.distribution_predictor(Transform=False)
data = Play.shape_data.create_residuals(predictor[0],lat_idx= 16,lon_idx = 32,min_run_idx=5, max_run_idx=11,Transform=False)
res = data.mrsol_res.values.flatten()

obs = data.mrsol.values.flatten()
obs = obs[np.isfinite(res)]
tas = data.tas.values.flatten()
tas = tas[np.isfinite(res)]
pr = data.pr.values.flatten()
pr = pr[np.isfinite(res)]

res = res[np.isfinite(res)]
print(tas.shape)
print(pr.shape)
print(obs.shape)
print(res.shape)
print(tas)
print(pr)
print(obs)
print(res)

var = predictor[1].predict(np.array([tas,pr]).T)

sigma = np.sqrt(np.maximum(var,1e-6))
z = res /sigma

crps = sigma * (
    z * (2 * norm.cdf(z) - 1)
    + 2 * norm.pdf(z)
    - 1 / np.sqrt(np.pi)
    )

print(f"model_crps mean: {crps.mean()}")
print(f"mu_crps mean: {abs(res).mean()}")
```
