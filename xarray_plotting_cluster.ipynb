```python
# Earth's average radius in meters
R = 6.371e6

# Coordinate spacing for this dataset is 2.5 x 2.5 degrees
dϕ = np.deg2rad(2.5)
dλ = np.deg2rad(2.5)

dlat_tas = R * dϕ * xr.ones_like(ds_tas.tas.lon) #xr.ones_like gibt  ein Objekt zurück wie das in den Klammern aber gefüllt mit 1ern.
dlon_tas = R * dλ * np.cos(np.deg2rad(ds_tas.tas.lat)) #np.cos... gibt wieder metamässig gleichen XArray
dlon_tas.name = "dlon"
dlat_tas.name = "dlat"

# returns an xarray DataArray!
#np.cos(np.deg2rad(ds_tas.lat))

cell_area_tas = dlon_tas * dlat_tas
cell_area_tas

#etwas automatischer (2.5 grad nicht von hand)
#dlon_tas_auto = R * np.gradient(np.deg2rad(ds_tas.tas.lon))
#dlat_tas_auto = R * np.gradient(np.deg2rad(ds_tas.tas.lat)) * np.cos(np.deg2rad(ds_tas.tas.lat))
#cell_area_tas_auto =  dlon_tas_auto * dlat_tas_auto 
#dlat_tas_auto

#da die anderen Variablen die gleichen degrees haben, kann man cell_area_tas für alle variablen als Gewicht verwendet werden 
```

```python
#Zeitreihe
ds_tas.tas.weighted(cell_area_tas).mean(["lon", "lat"]).plot(x="time");
plt.title("temparature weighted")
plt.show()
ds_tas.tas.mean(["lon", "lat"]).plot(x="time");
plt.title(" temparatur non-weighted, fehlersichtbar durch zu kalt")
plt.show()
ds_pr.pr.weighted(cell_area_tas).mean(["lon", "lat"]).plot(x="time");
plt.title("precipitation weighted")
plt.show()
ds_rsds.rsds.weighted(cell_area_tas).mean(["lon", "lat"]).plot(x="time");
plt.title("radiation weighted")
plt.show()
ds_hurs.hurs.weighted(cell_area_tas).mean(["lon", "lat"]).plot(x="time");
plt.title("relative humidity weighted")
plt.show()
```

```python
ds_tas = xr.load_dataset(Path('/net/atmos/data/cmip6-ng/tas/mon/g025/tas_mon_MPI-ESM1-2-LR_historical_r50i1p1f1_g025.nc'))

ds_tas_mon = ds_tas.sel(time=slice("1895-01-01", "1900-01-01")).resample(time="ME").mean()
ds_tas_mon = ds_tas_mon.sel(time= ds_tas_mon.time.dt.month == 7)
ds_tas_mon.tas.plot(col="time", col_wrap = 4, x = "lon");
plt.title("temparature in decade steps") #warum macht der nichts?
plt.show()
```

```python
ds_tas_dec = ds_tas.resample(time="10YE").mean()
#Auf folgender Zeitreihe sieht man nichts, da die Skala zu grob ist
ds_tas_dec.tas.plot(col="time", col_wrap = 4, x = "lon");
plt.title("temparature in decade steps") #warum macht der nichts?
plt.show()


#Hier ist die Differenz zu 1850ern gezeigt:
ds_tas_dec_diff = ds_tas_dec - ds_tas_dec.isel(time=0) #Jahrzehnt 1850
ds_tas_dec_diff.tas.plot(col="time", col_wrap = 4, x = "lon");  
plt.title("temparature in decade steps")
plt.show()
```
